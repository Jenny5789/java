## 💡 서블릿의 본질과 역할
서블릿은 **순수 자바**로 작성된 클래스입니다. 웹 브라우저가 HTTP 프로토콜로 보낸 요청을 받아들이고, 서버에서 동적으로 HTML 콘텐츠를 생성하여 응답하는 역할을 합니다.

*   **핵심 특징**
    *   **플랫폼 독립적:** 자바의 장점(객체지향, GC, 멀티스레딩)을 그대로 계승
    *   **컨테이너 의존적:** Tomcat, Jetty 등 WAS 안에서만 동작
    *   **스레드 기반:** 동시 요청 시 별도의 스레드로 병렬 처리
*   **동적 콘텐츠 예시**
    *   **요청:** "오늘 서울 날씨 알려줘" ☁️
    *   **처리:** DB 조회 🔍 → "맑음, 22도" → HTML 포맷팅 📝
    *   **응답:** 브라우저에 최종 결과 전송 📤

---

## 🌳 서블릿 API 계층 구조
서블릿은 계층적인 구조를 통해 기능을 확장해 나갑니다.

> **Servlet (인터페이스)**  
>   ↓ *implements*  
> **GenericServlet (추상클래스)**  
>   ↓ *extends*  
> **HttpServlet (우리가 상속받는 본체! ⭐)**

### 🛠️ HttpServlet 핵심 메서드
*   `service()`: 모든 요청의 **진입점**
*   `doGet()`: GET 방식 처리 (주로 페이지 조회)
*   `doPost()`: POST 방식 처리 (주로 데이터 전송/저장)
*   `doPut()` / `doDelete()`: 데이터 수정 및 삭제 처리

---

## 🔄 서블릿 생명주기 (Life Cycle)
서블릿은 톰캣에 의해 관리되며 정해진 순서대로 작동합니다.

**[ 🏁 Tomcat 시작 ]**
1.  **생성자 (1회):** 서블릿 인스턴스(객체) 생성 🏗️
2.  **init() (1회):** DB 연결 등 초기 설정 수행 ⚙️
3.  **service() (요청마다):** `doGet()` 또는 `doPost()` 호출 🏃

**[ 🛑 Tomcat 종료 ]**
4.  **destroy() (1회):** 사용하던 자원 정리 및 종료 🧹

## 📍 @WebServlet 어노테이션
> **"서블릿의 집주소를 알려주는 자동 등록 시스템"**

### ✨ 4단계 실행 과정
1.  **Step 1 (코드 작성):** `@WebServlet("/first")` 주소 설정 ✍️
2.  **Step 2 (컴파일):** `Ctrl+S` → `.java`가 `.class`로 변환 ⚙️
3.  **Step 3 (Tomcat 부팅):** `@WebServlet` 정보를 찾아 `/first` 주소 자동 등록 🔍
4.  **Step 4 (요청):** 브라우저가 주소 호출 → 매핑 테이블 확인 → 서블릿 실행 🚀

---

## ⚙️ 컴파일 과정
컴파일은 사람이 쓰는 언어를 컴퓨터가 이해하는 언어로 바꾸는 과정입니다.

*   **도구:** Eclipse에서 저장(`Ctrl+S`) 시 자동 수행
*   **결과:** `target/classes/FirstServlet.class` 생성
*   **참고:** `@WebServlet` 같은 설정값도 `.class` 파일에 담깁니다.

---

## 📐 클래스 vs 객체
*   **클래스(Class):** 붕어빵 틀 (설계도 - `FirstServlet.java`)
*   **객체(Object):** 실제 붕어빵 (실체 - `new FirstServlet()`)
*   **특징:** 우리가 직접 `new`를 하지 않아도, **Tomcat이 자동으로 객체를 생성**해 줍니다.

---

## 🔌 네트워크 포트 (Port)
서버 내에서 어떤 서비스(프로그램)를 찾아갈지 결정하는 번호입니다.
*   **80:** HTTP 기본 포트 (네이버 등 실제 서비스용, 생략 가능)
*   **8080:** **Tomcat 개발 기본 포트** (우리 실습용 주소)
*   **설정:** Eclipse → Servers → Tomcat 더블클릭 → Ports에서 확인 및 변경

---

## 🌊 전체 실행 흐름도
1.  **작성:** `FirstServlet.java` 작성 (어노테이션 포함)
2.  **컴파일:** 저장 시 `.class` 파일 자동 생성
3.  **부팅:** Tomcat 시작 시 주소(`/first`) 자동 등록
4.  **호출:** 브라우저에서 `localhost:8080/first` 요청
5.  **매핑:** 톰캣이 주소를 보고 해당 서블릿의 `doGet()` 실행
6.  **응답:** HTML 결과를 브라우저 화면에 출력
