###🔹 배열의 크기, length 필드

###🔹 배열과 for-each 문
- for - each 문 : 배열이나 나열의 각 원소를 순차적을 ㅗ접근하는데 유용
    1. Length 넘어가는 것을 방지(Index Out of Bounds) 방지
    2. 코드 가독성 향상
    3. 원본 데이터의 안전한 보호


       - 객체(몸)은 하나, 참조변수(팔)이 두 개인 상태.


       - 팔의 주소는 하나(몸)에서 나옴
       - 원래 데이터에 연결된 팔은 수정 가능하지만, 새로 만들어진 팔은 수정이 불가
       - 이를 통해 데이터의 안전한 보호가 가능

````python
public class Main {
    public static void main(String[] args) {
        int[] num= {1,2,3,4,5 };
        int sum = 0;
        for (int k : num)
            sum += k;
        System.out.println("합은" + sum);
    }
}

// int: 1차원 배열 형태
````

````python
String names[] = { "사과", "배", "바나나", "딸기", "포도" } ;
for (String s : names)
    System.out.print(s + " ");

// String:2차원 배열 형태
````

````python
enumWeek { 월, 화, 수, 목, 금, 토, 일}
for (Week day : Week.values()) // 반복될때마다day는월, 화, 수, 목, 금, 토, 일로설정
    System.out.print(day + "요일")

// enum:열거형
````

###🔹비정방형 배열, length
- 정방형 배열: 각 행의 열의 개수가 같은 배열


```
int i[][];
i = new int[4][4];    
```



- 비정방형 배열: 각 행의 열의 개수가 다른 배열


```
int i[][];
i = new int[4][];

i[0] = new int[1];  
i[1] = new int[2];
i[2] = new int[3];
i[3] = new int[4];
```


```
i.length: 2차원 배열의 행의 개수-> 4개
i[0].length: 0번째 행의 열의 개수-> 1
i[1].length: 1번째 행의 열의 개수-> 2
i[2].length: 2번째 행의 열의 개수-> 3
i[3].length: 3번째 행의 열의 개수-> 4
```

###🔹main() method

> public static void main(String[] args){
}
- public : 어느 위치에서나 호출 가능하도록 하기 위함
- static: 객체를 생성하지 않아도 프로그램에 올라갈 수 있도록 함
- void: return이 없음. main이 종료되면 프로그램 전체 종료.
- main: main method 로 호출
- String[]: 문자열 배열(콘솔로 우리가 문자를 입력함. 공백도 문자, 숫자도 문자)
- args: 매개변수(arguments)
