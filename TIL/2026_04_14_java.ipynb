### 🔹컴파일 오류

>  말, 문법이 안 통함

### 🔹논리적 오류


> 말도 잘 듣고 행동도 하는데 결과가 엉뚱함

    - 산술 및 게산 관련 오류, 조건문 및 반복문 로직 오류, 변수 초기화 및 공유 오류


### 🔹런타임 오류


> 말은 통하는데 행동하다 사고 침(프로그램 중단)


    - ArithmeticException(정수를 0으로 나눌 때 발생)
    - NullPointerException(null 레퍼런스를 참조할 때 발생)
    - IOException(입출력 동작 실패 또는 인터럽트 시 발생)
    - ArrayIndexOutOfBoundsException(배열의 범위를 벗어난 접근 시 발생)
    - NumberFormatException(문자열이 나타내는 숫자와 일치하지 않는 타입의 숫자로 변환 시 발생)
    - InputMismatchException(Scanner 클래스를 이용해 입력을 받을 때, 기대하는
     데이터 타입과 사용자가 실제로 입력한 타입이 일치하지 않을 경우 발생하는 런타임 예외.

### 🔹 예외 처리, try ~ catch ~ finally 문
    - finally 블록은 생략 가능
    - try 블록에서 예외가 발생하지 않는 정상적인 경우(try블록 -> finally 블록 문)
    - try 블록에서 예외가 발생한 경우(try 블록-> catch 문 -> finally 블록 문)

````python
try {
    예외가 발생할 가능성이 있는 실행문(try 블록);
}
catch{
    예외 처리문(catch 블록);
    ...
}
finally{
    예외 발생 여부와 상관없이 무조건 실행되는 문장(finally 블록);
    ...
}
````

# 📚 클래스와 객체

## 📝 객체 지향 프로그램(Object Oriented Program)

### 🔹 클래스와 객체
    - 클래스 : 설계도, 정의 명세, 객체의 속성(함수와) 행동(기능)을 가짐
        > ex) 붕어빵 틀, 사람

    - 객체 : 찍어낸 실체, 메모리 공간을 갖는 구체적인 실체, 인스턴스라고도 부름
        > ex) 붕어빵, 이재용, 정의선

    

```
#
public class Circle{  // public:접근 권한, class: 클래스 선언,  Circle: 클래스 이름

    public int radius; // 원의 반지름 필드
    public String name; //  원의 이름 필드
    // 위 2줄 = 필드(변수)

    public Circle(){  // 원의 생성자 메소드
    }
    public double getArea(){
        return 3.14 * radius * radius: // 원의 면적 계산 메소드
    }
    // 원의 생성자 메소드 ~ 원의 면적 계산 메소드 : 메소드
}
```



### 🔹 캡슐화

        - 객체를 캡슐로 싸서 내부를 볼 수 없게 하는 것
        - 외부로부터 객체 보호, 외부 접근 제한
        - 보안 및 간섭 방지

### 🔹 상속
    - 상위 객체의 속성이 하위 객체에 물려짐
    - 코드 재사용, 기능확장
    - 부모 클래스: 슈퍼 클래스
    - 자식 클래스: 서브 클래스, 슈퍼 클래스 재사용하고 새로운 특성 추가

### 🔹 다형성
    - 같은 이름의 메소드가 클래스나 객체에 따라 다르게 동작하도록 구현
    - 하나를 여러가지 형태로 사용 및 재정의

> Overloading : '다중 정의'

    > 같은 기능을 수행하지만, 입력 받는 데이터의 종류가 다양할 때 사용.

    > 같은 이름이지만 다르게 작동하는 여러 메소드

        > 인자 개수 type에 따라 다르게 호출됨

> Overriding : '재정의'

    > 슈퍼클래스의 메소드를 서브 클래스마다 다르게 구현
    > 기능을 필요에 따라 재정의해서 사용.
    


### 🔹추상화
    - 복잡한 세부 사항은 숨기고 공통적인 핵심만 추출함

    - 설계의 단순화
    

### 🔹생성자
- 생성자의 목적은 객체 "초기화"
- 생성자 이름은 클래스의 이름과 반드시 동일
- 리턴(= 반환) 타임(void) 없음
- new를 통해 객체를 생성할 때 , 객체 당 한 번 호출됨
- 생성자는 여러개 작성 가능(오버 로딩)
- 클래스에 생성자가 하나도 선언되지 않은 경우, 컴파일러에 의해 자동으로 삽입
    > ex) 생성자가 아예 없니?-> 내가 하나 넣어줄게 // 생성자를 직접 만들었구나? -> 난 개입하지 않을게.

### 🔹 this()
- 나 자신의 주소값을 통째로 넘길 때, 다른 생성자 호출(코드 중복 제거)
- 반드시 첫 번째 줄에서만 사용 가능함
- 생성자 내에서만 사용 가능


! this :  현재 객체의 주소값 그 자체
! this() : 나 자신의 또 다른 생성자르 부르는 리모콘(생성자 내부에서만 씀)

### 🔹 객체 소멸과 garbage collection
- 객체 소멸 : new에 할당된 객체 메모리를 자바 가상 기계의 가용 메모리로 되돌려 주는 행위
- garbage : 누구도 사용할 수 없게 된 메모리, 자바에서 자동 회수하는 과정


## 📝 접근 지정자

### 🔹 접근지정자: private, protected, public, default

    > 목적 - 클래스나 일부 멤버를 공개하여 다른 클래스에서 접근하도록 허용
         - 객체 지향 언어의 캡슐화 정책은 멤버를 보호하는 것


### 🔹 public
    - 패키지에 관계없이 모든 클래스에 접근 허용

### 🔹 private
    - 동일 클래스 내에만 접근 허용
    - 상속 받은 서브 클래스에서 접근 불가

### 🔹 protected
    - 같은 패키지 내의 다른 모든 클래스에게 접근 허용
    - 상속 받은 서브 클래스는 다른 패키지에 있어도 접근 가능
    

### 🔹 default
    - 같은 패키지 내의 다른 클래스에게 접근 허용

### 🔹 static & non-static
- static: 클래스 소유, 공유 변수, 객체 생성 없이 사용 가능
- non-static : 객체 소유, 개별 변수, 반드시 new 후 사용 가능
