###🔹printf 서식 지정자와 자료형 매칭

> %d --> int, long, short, byte

> %f --> float, double

ex)
int won = scanner.nextInt();

double dollar = won / 1490;

System.out.printf("**%d원은 $%.2f**입니다.", won, dollar);



*  won 입력 받아->  %d(=  정수 출력)
*  dollar --> %.2f  = 소수점 두째 자리까지 출력





---



💻 실습 코드

````python
package ex01;
import java.util.Scanner;
public class extra {

	public static void main(String[] args) {
		Scanner scanner = new Scanner(System.in);
		System.out.println("원화를 입력하세요(단위: 원) >> ");

		int won = scanner.nextInt();
		double dollar = won / 1490;

		System.out.printf("%d원은 $%.2f입니다.", won, dollar);
		scanner.close();
	}
}
````

## 📚 JAVA 기초 2 _ 반복문

###🔹for문


```
#  for(초기문; 조건식; 반복 후 작업){
    ...작업문...
    }
```


```
# for(초기문; 첫번째; 세번째){
    ...두번째...
    }
```
- for문 : 초기문 -> 첫번째-> 두번째-> 세번째 순으로 진행.

- 실행해야할 횟수를 알고 있을 때 사용.

- '초기문'으로 값 초기화 후, '조건식' 으로 진행여부 판단. 만족 시 '작업문' 실행 후 '반복 후 작업' 진행.




```
💻 예시
int i;
for(i =0; i < 10 ; i ++){
    System.out.print(i);
}
```
--> i는 정수

--> i가 0으로 초기화함.  'i < 10' 조건식 만족 --> 'print()'작업문  실행 후 반복 후 작업인 'i++'를 한다.

--> 0123456789

###🔹while문
```
#  while(조건식){
    ...작업문...
    }
```


```
# while(첫번째){
    ...두번째...
    }
```
- while 문 :첫번째-> 두번째 순으로 진행.

- 실행해야할 횟수를 모른다. 조건 만족 여부에 따라 진행됨



```
💻 예시
i = 0;
while(i < 10 ){
    System.out.print(i);
    i++;
}
```

--> i는 0

--> 'i< 10' 조건식 만족-> 'i'프린트-> 'i++'

--> 0123456789

###🔹do while문
```
#  do{
    ...작업문...
    }while(조건식);
```


```
#  do{
    ...첫번째...
    }while(두번째);
```
- do~ while 문 :첫번째-> 두번째 순으로 진행.

- 무조건 작업문은 최소 한 번 실행.


```
💻 예시
i = 0;
do {
    System.out.print(i);
    i++
} while(i<10);
```

--> i는 0

--> 작업문 'print i' 후 i++

--> 'i<10' 할때까지 반복

--> i가 10이면 끝남
